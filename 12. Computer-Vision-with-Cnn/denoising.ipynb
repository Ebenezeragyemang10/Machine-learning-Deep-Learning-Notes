```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy
import os
from pathlib import Path
```

```python
data_dir = Path("../dataset/Denoising-data/train")
files = sorted(data_dir.glob("*.npz"))

print(f"number of files: {len(files)}")
print(files[0])
np.load(files[0])
```

```python
for file in files:
    npz = np.load(file)
    
    
    print("\n", file.name)
    print("keys", npz.files)
    for key in npz.files:
        print(
            f'{key} \n' 
            f'shape: {npz[key].shape}'
            f"dtype={npz[key].dtype}"
        )
    
```

```python
from random import randint,seed
seed(42)
rand_no = randint(0, len(files)) 
random_npz = np.load(files[rand_no])

data = random_npz["data"]
itp = int(random_npz["itp"])

print(itp)
```

```python
fig, ax = plt.subplots(3, 1, figsize=(10, 10))
for i in range(3):
    ax[i].plot(data[:, i])
    ax[i].axvline(
        itp, 
        linestyle = "--",
        label="p-wave arival",
        color="r"
    )
    ax[i].set_ylabel(f'channel {i+1}')
    ax[i].legend()
ax[-1].set_xlabel("Sample")

plt.tight_layout()
plt.show()
    
```
